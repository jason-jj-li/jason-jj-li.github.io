## Part 2: Population Pyramids and Age Structure

### 📊 What is a Population Pyramid?

A **Population Pyramid** is one of the most fundamental and important visualization tools in demography. It displays population distribution through horizontal bar charts showing:

- **Left Side**: Male population (by age group)
- **Right Side**: Female population (by age group)
- **Vertical Axis**: Age groups (typically 5-year intervals)
- **Horizontal Axis**: Population numbers or percentages

### 🔍 Four Basic Population Pyramid Shapes

#### 1. Expansive (Pyramid Shape)
- **Characteristics**: Wide base, narrow top
- **Meaning**: High birth rates, young population
- **Typical Countries**: Nigeria, Pakistan, Congo
- **Policy Implications**: Requires significant investment in education and employment

#### 2. Stationary (Bell Shape)
- **Characteristics**: Relatively balanced across age groups
- **Meaning**: Balanced birth and death rates
- **Typical Countries**: United States, Australia
- **Policy Implications**: Balanced social security needs

#### 3. Constrictive (Pear Shape)
- **Characteristics**: Narrowing base, expanding middle
- **Meaning**: Low birth rates, aging population
- **Typical Countries**: China, Brazil, South Korea
- **Policy Implications**: Increasing pension pressure

#### 4. Inverted (Inverted Pyramid)
- **Characteristics**: Very narrow base, wide top
- **Meaning**: Extremely low birth rates, super-aged society
- **Typical Countries**: Japan, Germany, Italy
- **Policy Implications**: Population sustainability crisis

````python
# Create sample data for four different pyramid shapes

age_groups = ['0-4', '5-9', '10-14', '15-19', '20-24', '25-29', 
              '30-34', '35-39', '40-44', '45-49', '50-54', '55-59',
              '60-64', '65-69', '70-74', '75-79', '80+']

# 1. Expansive: Developing country (e.g., Nigeria)
expansive_male = [120, 115, 110, 105, 100, 95, 90, 85, 75, 65, 55, 45, 35, 25, 18, 12, 8]
expansive_female = [115, 110, 105, 100, 95, 90, 85, 80, 72, 62, 52, 42, 32, 23, 17, 11, 9]

# 2. Stationary: Developed country (e.g., USA)
stationary_male = [65, 68, 70, 72, 75, 78, 80, 82, 85, 87, 85, 80, 75, 65, 50, 35, 20]
stationary_female = [62, 65, 67, 69, 72, 75, 77, 79, 82, 84, 82, 77, 72, 62, 48, 33, 22]

# 3. Constrictive: Transitioning country (e.g., China)
constrictive_male = [55, 52, 50, 65, 85, 95, 100, 105, 110, 105, 95, 85, 70, 55, 40, 28, 15]
constrictive_female = [52, 50, 48, 62, 82, 92, 97, 102, 107, 102, 92, 82, 67, 52, 38, 26, 17]

# 4. Inverted: Super-aged country (e.g., Japan)
inverted_male = [45, 43, 42, 48, 60, 72, 85, 95, 100, 105, 108, 105, 98, 88, 75, 55, 35]
inverted_female = [43, 41, 40, 46, 58, 70, 83, 93, 98, 103, 106, 103, 96, 86, 73, 53, 38]

# Create DataFrame
pyramid_data = pd.DataFrame({
    'Age_Group': age_groups,
    'Expansive_M': expansive_male,
    'Expansive_F': expansive_female,
    'Stationary_M': stationary_male,
    'Stationary_F': stationary_female,
    'Constrictive_M': constrictive_male,
    'Constrictive_F': constrictive_female,
    'Inverted_M': inverted_male,
    'Inverted_F': inverted_female
})

print("📊 Population Pyramid Data Preview (in thousands)")
print("="*70)
print(pyramid_data.head(10))
print("\n✅ Data created successfully! Includes 4 population structure types")
````

````python
# Plot all four population pyramid types for comparison

fig, axes = plt.subplots(2, 2, figsize=(16, 14))
fig.suptitle('Four Types of Population Pyramids', 
             fontsize=16, fontweight='bold', y=0.995)

pyramid_types = [
    ('Expansive', 'Expansive (Nigeria-type)', 0, 0, 'steelblue', 'crimson'),
    ('Stationary', 'Stationary (USA-type)', 0, 1, '#2ecc71', '#e74c3c'),
    ('Constrictive', 'Constrictive (China-type)', 1, 0, '#3498db', '#e67e22'),
    ('Inverted', 'Inverted (Japan-type)', 1, 1, '#9b59b6', '#e91e63')
]

for ptype, title, row, col, color_m, color_f in pyramid_types:
    ax = axes[row, col]
    
    # Get data
    male_data = -pyramid_data[f'{ptype}_M']  # Negative for left side
    female_data = pyramid_data[f'{ptype}_F']
    
    # Plot bars
    ax.barh(pyramid_data['Age_Group'], male_data, 
            label='Male', color=color_m, alpha=0.8, edgecolor='black', linewidth=0.5)
    ax.barh(pyramid_data['Age_Group'], female_data, 
            label='Female', color=color_f, alpha=0.8, edgecolor='black', linewidth=0.5)
    
    # Add center line
    ax.axvline(x=0, color='black', linewidth=2)
    
    # Set title and labels
    ax.set_title(title, fontsize=13, fontweight='bold', pad=10)
    ax.set_xlabel('Population (thousands)', fontsize=10, fontweight='bold')
    ax.set_ylabel('Age Group', fontsize=10, fontweight='bold')
    
    # Set x-axis ticks
    ticks = [-120, -80, -40, 0, 40, 80, 120]
    ax.set_xticks(ticks)
    ax.set_xticklabels([abs(t) for t in ticks])
    
    # Add grid
    ax.grid(True, alpha=0.3, axis='x')
    ax.legend(loc='upper right', fontsize=9)
    
    # Add text annotations
    ax.text(-60, 15, 'MALE', fontsize=11, fontweight='bold', 
            ha='center', va='center', bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))
    ax.text(60, 15, 'FEMALE', fontsize=11, fontweight='bold', 
            ha='center', va='center', bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

plt.tight_layout()
plt.show()

print("\n✅ Four population pyramids generated!")
print("\n💡 Key Observations:")
print("  • Expansive: Widest base, indicates high birth rate")
print("  • Stationary: Relatively balanced, indicates stable population")
print("  • Constrictive: Expanding middle, narrowing base")
print("  • Inverted: Top-heavy, severe aging")
````

### 📈 Age Structure Analysis

Let's calculate key indicators for each population structure:

````python
# Calculate age structure indicators for each type

def calculate_age_structure(male_pop, female_pop, age_groups):
    """
    Calculate population age structure indicators
    """
    total_pop = male_pop + female_pop
    
    # Age group divisions (assuming first 3 groups are 0-14, middle 11 groups are 15-64, last 3 groups are 65+)
    youth = total_pop[:3].sum()  # 0-14 years
    working_age = total_pop[3:14].sum()  # 15-64 years
    elderly = total_pop[14:].sum()  # 65+ years
    total = total_pop.sum()
    
    # Calculate percentages
    youth_pct = (youth / total) * 100
    working_pct = (working_age / total) * 100
    elderly_pct = (elderly / total) * 100
    
    # Calculate dependency ratios
    youth_dependency = (youth / working_age) * 100
    age_dependency = (elderly / working_age) * 100
    total_dependency = youth_dependency + age_dependency
    
    # Calculate sex ratio (male/female × 100)
    sex_ratio = (male_pop.sum() / female_pop.sum()) * 100
    
    # Calculate median age (simplified estimate)
    cumsum = np.cumsum(total_pop)
    median_idx = np.searchsorted(cumsum, total / 2)
    median_age = median_idx * 5 + 2.5  # Simplified estimate
    
    return {
        '0-14 (%)': youth_pct,
        '15-64 (%)': working_pct,
        '65+ (%)': elderly_pct,
        'Youth Dependency': youth_dependency,
        'Age Dependency': age_dependency,
        'Total Dependency': total_dependency,
        'Sex Ratio': sex_ratio,
        'Median Age': median_age,
        'Total Pop (K)': total
    }

# Calculate indicators for each type
results = {}
for ptype in ['Expansive', 'Stationary', 'Constrictive', 'Inverted']:
    male = pyramid_data[f'{ptype}_M'].values
    female = pyramid_data[f'{ptype}_F'].values
    results[ptype] = calculate_age_structure(male, female, age_groups)

# Create results DataFrame
results_df = pd.DataFrame(results).T

print("\n" + "="*90)
print("📊 Key Indicators Comparison of Four Population Structures")
print("="*90)
print(results_df.round(2))

print("\n\n💡 Indicator Interpretation:")
print("-" * 90)
print("\n1. Expansive (Developing Countries):")
print("   • Highest 0-14 age proportion → Great future growth potential")
print("   • High youth dependency → Large education investment needs")
print("   • Low median age → Young society")

print("\n2. Stationary (Developed Countries):")
print("   • Relatively balanced age structure")
print("   • Moderate total dependency ratio")
print("   • Relatively stable population")

print("\n3. Constrictive (Transitioning Countries):")
print("   • High working-age population proportion → 'Demographic dividend' period")
print("   • But clear aging trend")
print("   • Need to prepare for future aging")

print("\n4. Inverted (Super-aged Countries):")
print("   • Highest 65+ proportion → Severe aging")
print("   • Extremely high age dependency → Huge pension pressure")
print("   • Population sustainability faces challenges")
````

## Part 3: Core Demographic Indicators

### 📊 Basic Demographic Rates

The three most fundamental indicators in demography:

#### 1. Crude Birth Rate (CBR)
```
CBR = (Annual Births / Mid-year Total Population) × 1,000
```
- High CBR (>30): Developing countries
- Moderate CBR (15-30): Transitioning countries  
- Low CBR (<15): Developed countries

#### 2. Crude Death Rate (CDR)
```
CDR = (Annual Deaths / Mid-year Total Population) × 1,000
```
Note: Aging countries naturally have higher CDR

#### 3. Natural Increase Rate (NIR)
```
NIR = CBR - CDR (per thousand)
or
% Growth = ((Current Year Pop - Previous Year Pop) / Previous Year Pop) × 100
```

### Real-World Case: Multi-Country Population Dynamics Comparison

````python
# Create multi-country real demographic data (based on 2023 actual data)

countries_vital_stats = {
    'Country': ['Nigeria', 'India', 'Brazil', 'China', 'Japan', 'Germany'],
    'Population (M)': [223, 1417, 215, 1425, 125, 84],
    'Births (K)': [8140, 25079, 3010, 11400, 788, 731],
    'Deaths (K)': [2051, 9919, 1613, 12540, 1438, 1059],
    'TFR': [4.9, 2.1, 1.7, 1.3, 1.2, 1.5],
    'Life Expectancy': [54.7, 70.3, 76.7, 77.5, 85.0, 81.8],
    'Median Age': [18.5, 28.1, 34.8, 38.5, 48.8, 48.0],
    'Urbanization (%)': [52, 35, 88, 64, 92, 77]
}

df_vital = pd.DataFrame(countries_vital_stats)

# Calculate demographic rates
df_vital['CBR'] = (df_vital['Births (K)'] / df_vital['Population (M)'])
df_vital['CDR'] = (df_vital['Deaths (K)'] / df_vital['Population (M)'])
df_vital['NIR'] = df_vital['CBR'] - df_vital['CDR']
df_vital['Growth (%)'] = (df_vital['NIR'] / 10).round(2)

print("\n" + "="*100)
print("🌍 Six-Country Demographic Indicators Comparison (2023 Data)")
print("="*100)

# Display basic data
print("\n📋 Basic Population Data:")
display_cols = ['Country', 'Population (M)', 'Births (K)', 'Deaths (K)']
print(df_vital[display_cols].to_string(index=False))

# Display calculated rates
print("\n\n📊 Calculated Demographic Rates:")
rate_cols = ['Country', 'CBR', 'CDR', 'NIR', 'Growth (%)']
print(df_vital[rate_cols].round(2).to_string(index=False))

# Display development indicators
print("\n\n📈 Social Development Indicators:")
dev_cols = ['Country', 'TFR', 'Life Expectancy', 'Median Age', 'Urbanization (%)']
print(df_vital[dev_cols].round(1).to_string(index=False))
````

````python
# Visualize demographic rates comparison

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Key Demographic Indicators Across Six Countries', 
             fontsize=15, fontweight='bold')

countries = df_vital['Country'].values

# 1. Birth Rate vs Death Rate
ax1 = axes[0, 0]
x = np.arange(len(countries))
width = 0.35
ax1.bar(x - width/2, df_vital['CBR'], width, label='Birth Rate (CBR)', 
        color='#2ecc71', alpha=0.8, edgecolor='black')
ax1.bar(x + width/2, df_vital['CDR'], width, label='Death Rate (CDR)', 
        color='#e74c3c', alpha=0.8, edgecolor='black')
ax1.set_ylabel('Rate (per 1,000)', fontweight='bold')
ax1.set_title('Birth Rate vs Death Rate', fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(countries, rotation=45, ha='right')
ax1.legend()
ax1.grid(True, alpha=0.3, axis='y')

# 2. Natural Increase Rate
ax2 = axes[0, 1]
colors_growth = ['#e74c3c' if x > 20 else '#f39c12' if x > 5 else '#2ecc71' if x > 0 else '#3498db' 
                 for x in df_vital['NIR']]
ax2.barh(countries, df_vital['NIR'], color=colors_growth, alpha=0.8, edgecolor='black')
ax2.set_xlabel('Natural Increase Rate (per 1,000)', fontweight='bold')
ax2.set_title('Population Natural Increase Rate', fontweight='bold')
ax2.axvline(x=0, color='black', linewidth=1.5)
ax2.grid(True, alpha=0.3, axis='x')
for i, v in enumerate(df_vital['NIR']):
    ax2.text(v + 1, i, f'{v:.1f}', va='center', fontweight='bold')

# 3. Total Fertility Rate
ax3 = axes[0, 2]
colors_tfr = ['#e74c3c' if x > 3.5 else '#f39c12' if x > 2.1 else '#3498db' 
              for x in df_vital['TFR']]
ax3.bar(countries, df_vital['TFR'], color=colors_tfr, alpha=0.8, edgecolor='black', linewidth=1.2)
ax3.axhline(y=2.1, color='red', linestyle='--', linewidth=2, label='Replacement Level 2.1')
ax3.set_ylabel('Children per Woman', fontweight='bold')
ax3.set_title('Total Fertility Rate (TFR)', fontweight='bold')
ax3.set_xticklabels(countries, rotation=45, ha='right')
ax3.legend()
ax3.grid(True, alpha=0.3, axis='y')
for i, v in enumerate(df_vital['TFR']):
    ax3.text(i, v + 0.15, f'{v:.1f}', ha='center', fontweight='bold')

# 4. Life Expectancy
ax4 = axes[1, 0]
colors_le = plt.cm.RdYlGn(df_vital['Life Expectancy'] / 90)
ax4.bar(countries, df_vital['Life Expectancy'], color=colors_le, alpha=0.8, edgecolor='black', linewidth=1.2)
ax4.set_ylabel('Age (years)', fontweight='bold')
ax4.set_title('Life Expectancy at Birth', fontweight='bold')
ax4.set_xticklabels(countries, rotation=45, ha='right')
ax4.set_ylim([0, 90])
ax4.grid(True, alpha=0.3, axis='y')
for i, v in enumerate(df_vital['Life Expectancy']):
    ax4.text(i, v + 2, f'{v:.1f}', ha='center', fontweight='bold')

# 5. Median Age
ax5 = axes[1, 1]
colors_age = ['#2ecc71' if x < 25 else '#f39c12' if x < 40 else '#e74c3c' 
              for x in df_vital['Median Age']]
ax5.bar(countries, df_vital['Median Age'], color=colors_age, alpha=0.8, edgecolor='black', linewidth=1.2)
ax5.set_ylabel('Age (years)', fontweight='bold')
ax5.set_title('Median Age (Aging Indicator)', fontweight='bold')
ax5.set_xticklabels(countries, rotation=45, ha='right')
ax5.grid(True, alpha=0.3, axis='y')
for i, v in enumerate(df_vital['Median Age']):
    ax5.text(i, v + 1.5, f'{v:.1f}', ha='center', fontweight='bold')

# 6. Population Size
ax6 = axes[1, 2]
colors_pop = plt.cm.Spectral(np.linspace(0, 1, len(countries)))
ax6.barh(countries, df_vital['Population (M)'], color=colors_pop, alpha=0.8, edgecolor='black', linewidth=1.2)
ax6.set_xlabel('Population (millions)', fontweight='bold')
ax6.set_title('Total Population Size', fontweight='bold')
ax6.grid(True, alpha=0.3, axis='x')
for i, v in enumerate(df_vital['Population (M)']):
    ax6.text(v + 30, i, f'{v}M', va='center', fontweight='bold')

plt.tight_layout()
plt.show()

print("\n✅ Visualization complete!")
print("\n💡 Key Findings:")
print("  🔴 Nigeria: Extremely high growth rate (27‰), young population")
print("  🟡 India: Fertility rate near replacement level (2.1)")
print("  🟢 Brazil: In transition, fertility below replacement level")
print("  🔵 China: Low fertility (1.3), population about to decline")
print("  🟣 Japan: Super-aged society, population already declining")
print("  🟠 Germany: Low fertility, relies on immigration to maintain population")
````

### 📐 Formula Summary

| Indicator | Formula | Unit |
|-----------|---------|------|
| **Crude Birth Rate (CBR)** | (Births / Total Population) × 1,000 | ‰ |
| **Crude Death Rate (CDR)** | (Deaths / Total Population) × 1,000 | ‰ |
| **Natural Increase Rate (NIR)** | CBR - CDR | ‰ |
| **Growth Rate (%)** | NIR / 10 | % |
| **Total Fertility Rate (TFR)** | Average children per woman | Children/Woman |
| **Infant Mortality Rate (IMR)** | (Deaths under 1 year / Live Births) × 1,000 | ‰ |
| **Age-Specific Fertility Rate** | Σ(Fertility rates by age group) | - |
