## Part 2: Package Version Check

Run this cell. If it prints version numbers without errors, your environment is correctly set up.

````python
import sys
import importlib

required = {
    'pandas':      '2.0',
    'numpy':       '1.24',
    'scipy':       '1.10',
    'matplotlib':  '3.7',
    'seaborn':     '0.12',
    'statsmodels': '0.14',
}

print(f'Python : {sys.version.split()[0]}')
print()

all_ok = True
for pkg, min_ver in required.items():
    try:
        mod = importlib.import_module(pkg)
        ver = getattr(mod, '__version__', 'unknown')
        status = 'OK' if ver >= min_ver else f'WARN (need >={min_ver})'
        print(f'  {pkg:<14} {ver:<12} [{status}]')
        if 'WARN' in status:
            all_ok = False
    except ImportError:
        print(f'  {pkg:<14} NOT INSTALLED  [ERROR]')
        all_ok = False

print()
if all_ok:
    print('All packages installed and up to date. You are ready.')
else:
    print('Some packages are missing or outdated.')
    print('Run: pip install --upgrade ' + ' '.join(required.keys()))
````

## Part 4: Hello Simulation — Your First Run

This is the exact pattern every post in this series will follow.  
Run these three cells and confirm you see output.

````python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')

# Step 1: Set parameters we know are true
TRUE_MEAN = 170   # e.g. height in cm
TRUE_STD  = 8
N         = 300

# Step 2: Simulate data
rng  = np.random.default_rng(42)
data = rng.normal(loc=TRUE_MEAN, scale=TRUE_STD, size=N)

# Step 3: Estimate from data
est_mean = data.mean()
est_std  = data.std(ddof=1)

# Step 4: Verify
print('=== Hello Simulation ===')
print(f'True mean : {TRUE_MEAN}    | Estimated: {est_mean:.2f}  | Error: {abs(est_mean-TRUE_MEAN):.2f}')
print(f'True std  : {TRUE_STD}     | Estimated: {est_std:.2f}   | Error: {abs(est_std-TRUE_STD):.2f}')
print()
print('If you see numbers above — Python is working correctly.')
````

````python
# A simple plot — if this displays, matplotlib is working
x = np.linspace(data.min(), data.max(), 200)
from scipy import stats

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(data, bins=30, density=True, color='#2E86AB', edgecolor='white', alpha=0.8, label='Simulated data')
ax.plot(x, stats.norm.pdf(x, TRUE_MEAN, TRUE_STD),
        color='#E74C3C', linewidth=2.5, label=f'True distribution (mean={TRUE_MEAN}, sd={TRUE_STD})')
ax.axvline(est_mean, color='#F18F01', linestyle='--', linewidth=2, label=f'Estimated mean = {est_mean:.1f}')
ax.set_xlabel('Value')
ax.set_ylabel('Density')
ax.set_title('Hello Simulation: estimated distribution vs. true distribution')
ax.legend()
plt.tight_layout()
plt.show()

print('If you see a histogram with a red curve — matplotlib is working correctly.')
````

````python
# Test pandas and statsmodels
import pandas as pd
import statsmodels.api as sm

df = pd.DataFrame({'height': data})
print('pandas DataFrame shape:', df.shape)
print(df.describe().round(2))
print()
print('If you see a summary table above — pandas is working correctly.')
````

## Common Installation Issues

| Problem | Solution |
|---|---|
| `python` command not found | Restart terminal; on macOS/Linux use `python3` |
| `pip` command not found | Try `pip3`; or run `python -m pip install ...` |
| `ModuleNotFoundError` | Re-run the pip install command and confirm no errors were printed |
| Jupyter opens wrong Python | Run `python -m jupyter lab` to ensure the correct Python's Jupyter is launched |
| `seaborn-v0_8-whitegrid` not found | Upgrade seaborn: `pip install --upgrade seaborn` |
| Plot doesn't display | Add `%matplotlib inline` at the top of your notebook |
